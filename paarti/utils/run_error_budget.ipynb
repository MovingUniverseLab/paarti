#### **The following error budget items are computed**: *all PSD+NCPA seen mode, ``sim.seed=1``*

**DM fitting error**: ``step1`` gives DM fitting error which depends on the actuator spacing.

**WFS aliasing error**: the quadrature difference between  ``step1`` and ``step2`` gives WFS aliasing error.

**Servo lag error**: the quadrature difference between ``step2b`` and ``step3`` gives servo lag error.

**WFS nonlinearity error**: the quadrature difference between ``step3`` and ``step4`` gives WFS non-linearity error.

**WFS noise effect**: the quadrature difference between ``step4`` and ``step5`` gives WFS noise error.

```python
# DM fitting error is pulled directly
tot_dm_fitting_error = tot_maos_err_step1
tt_dm_fitting_error = tt_maos_err_step1
ho_dm_fitting_error = ho_maos_err_step1
gc_tot_dm_fitting_error = gc_tot_maos_err_step1
gc_tt_dm_fitting_error = gc_tt_maos_err_step1
gc_ho_dm_fitting_error = gc_ho_maos_err_step1

# Quadrature diff between step1 and step2 for WFS aliasing error
tot_wfs_alias_error = np.sqrt(np.abs(tot_maos_err_step1**2 - tot_maos_err_step2**2))
tt_wfs_alias_error = np.sqrt(np.abs(tt_maos_err_step1**2 - tt_maos_err_step2**2))
ho_wfs_alias_error = np.sqrt(np.abs(ho_maos_err_step1**2 - ho_maos_err_step2**2))
gc_tot_wfs_alias_error = np.sqrt(np.abs(gc_tot_maos_err_step1**2 - gc_tot_maos_err_step2**2))
gc_tt_wfs_alias_error = np.sqrt(np.abs(gc_tt_maos_err_step1**2 - gc_tt_maos_err_step2**2))
gc_ho_wfs_alias_error = np.sqrt(np.abs(gc_ho_maos_err_step1**2 - gc_ho_maos_err_step2**2))

# Servo lag error is quad diff between step2 and step3 (technically step2b but that is an MCAO sim)
tot_servo_lag_error = np.sqrt(np.abs(tot_maos_err_step2**2 - tot_maos_err_step3**2))
tt_servo_lag_error = np.sqrt(np.abs(tt_maos_err_step2**2 - tt_maos_err_step3**2))
ho_servo_lag_error = np.sqrt(np.abs(ho_maos_err_step2**2 - ho_maos_err_step3**2))
gc_tot_servo_lag_error = np.sqrt(np.abs(gc_tot_maos_err_step2**2 - gc_tot_maos_err_step3**2))
gc_tt_servo_lag_error = np.sqrt(np.abs(gc_tt_maos_err_step2**2 - gc_tt_maos_err_step3**2))
gc_ho_servo_lag_error = np.sqrt(np.abs(gc_ho_maos_err_step2**2 - gc_ho_maos_err_step3**2))

# WFS non-linearity error is quad diff between step3 and step4
tot_wfs_non_lin_error = np.sqrt(np.abs(tot_maos_err_step3**2 - tot_maos_err_step4**2))
tt_wfs_non_lin_error = np.sqrt(np.abs(tt_maos_err_step3**2 - tt_maos_err_step4**2))
ho_wfs_non_lin_error = np.sqrt(np.abs(ho_maos_err_step3**2 - ho_maos_err_step4**2))
gc_tot_wfs_non_lin_error = np.sqrt(np.abs(gc_tot_maos_err_step3**2 - gc_tot_maos_err_step4**2))
gc_tt_wfs_non_lin_error = np.sqrt(np.abs(gc_tt_maos_err_step3**2 - gc_tt_maos_err_step4**2))
gc_ho_wfs_non_lin_error = np.sqrt(np.abs(gc_ho_maos_err_step3**2 - gc_ho_maos_err_step4**2))

# WFS noise error is quad diff between step4 and step5
tot_wfs_noise = np.sqrt(np.abs(tot_maos_err_step4**2 - tot_maos_err_step5**2))
tt_wfs_noise = np.sqrt(np.abs(tt_maos_err_step4**2 - tt_maos_err_step5**2))
ho_wfs_noise = np.sqrt(np.abs(ho_maos_err_step4**2 - ho_maos_err_step5**2))
gc_tot_wfs_noise = np.sqrt(np.abs(gc_tot_maos_err_step4**2 - gc_tot_maos_err_step5**2))
gc_tt_wfs_noise = np.sqrt(np.abs(gc_tt_maos_err_step4**2 - gc_tt_maos_err_step5**2))
gc_ho_wfs_noise = np.sqrt(np.abs(gc_ho_maos_err_step4**2 - gc_ho_maos_err_step5**2))
```
