```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.io import readsav
from pathlib import Path
import maos_utils as mu
from astropy.io import fits
```

```python
tt = mu.tt_residuals()
```

```python
tt_break_frequencies = tt[3]
```

```python
plt.rcParams.update({
    "text.usetex": False, # turn off latex rendering
    "font.sans-serif": "Helvetica",
})
```

```python
mean_freq = np.nanmean(tt_break_frequencies)
```

```python
fig, axes = plt.subplots(figsize=(7.5, 5.0))
plt.hist(tt_break_frequencies, bins=50, color="indigo", label="Frequency (Hz)")
plt.title("Tip-tilt 'break' frequencies for 2017 LGS telemetry streams", fontsize=14)
plt.xlabel("Frequency (Hz)", fontsize=12)
plt.ylabel("Counts", fontsize=12)
plt.legend(fontsize=8)
plt.tight_layout()
plt.grid(True, linestyle='dotted')
plt.axvline(mean_freq, color='k', 
            linestyle='dashed', linewidth=1)
plt.show()
```
