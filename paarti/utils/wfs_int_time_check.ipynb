```python
from astropy.io import fits
file = "/Users/bdigia/work/ao/airopa_input/20150514nirc2_kp/c1091.fits"
print(f"MAOS HO SHWFS int time is {(1.0/472.0)*1000.0} ms = STRAP according to maos_done.conf")
with fits.open(file) as fits_file:
    hdu = fits_file[0]
    psf = hdu.data
    hdr = hdu.header
    # STRAP WFS integration time (milli-sec)
    hdr_strap_int_time = float(hdr['STINTTIM'])
    # Frame rate for WFS cam (Hz) <-- from FITS header of file above
    hdr_shwfs_frame_rate = float(hdr['WSFRRT'])
    hdr_shwfs_int_time = (1.0/hdr_shwfs_frame_rate)*1000.0
print(f"HO SHWFS int time (ms) = {hdr_shwfs_int_time} | STRAP int time (ms) = {hdr_strap_int_time}")
print(f"HO SHWFS frame rate (Hz) = {hdr_shwfs_frame_rate} | STRAP frame rate (Hz) = {(1.0/(hdr_strap_int_time/1000.0))}")
```
